## Bloque 4: Lógica algorítmica y funciones personalizadas

```python
# Función para clasificar las emisiones totales de GEI
def clasificar_emisiones(emisiones):
    if pd.isna(emisiones):
        return "Sin dato"
    elif emisiones < 100:
        return "Emisiones bajas"
    elif emisiones < 1000:
        return "Emisiones medias"
    else:
        return "Emisiones altas"

# Se aplica la función a la columna total_ghg
df["categoria_emisiones"] = df["total_ghg"].apply(clasificar_emisiones)

# Se muestra resultado
display(df[["country", "year", "total_ghg", "categoria_emisiones"]].head(100000))
```

```python

#Función lambda para redondear el consumo energético

df["energy_per_capita_redondeada"] = df["energy_per_capita"].apply(lambda x: round(x, 2) if pd.notna(x) else x)

# Se muestra el resultado
display(df[["energy_per_capita", "energy_per_capita_redondeada"]].head(10))
```
